```python
import numpy as np
import matplotlib.pyplot as plt
ur_peak = 1.2
ur_phase = 0
ul_peak = 8.5
ul_phase = 500e-6
f = 50
w = 2 * np.pi * f
phi = -2 * np.pi * f * ul_phase

print(np.degrees(phi))

t = np.linspace(0, 2e-2, 100)

# lissajous

x = ul_peak * np.sin(w * t + phi)
y = ur_peak * np.sin(w * t)

plt.plot(t, x, t, y)
plt.show()
plt.grid()
plt.plot(x, y)
plt.show()
```
